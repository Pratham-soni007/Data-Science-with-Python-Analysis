```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=df,x="Price_USD",bins=30,kde=True)
plt.title("Distribution of Price in USD",
          fontsize=16,
          fontweight="bold",
          color="darkblue",
          pad=20)
plt.xlabel("Price in USD",fontsize=12,fontweight="bold",color="darkblue")
plt.ylabel("Count",fontsize=12,fontweight="bold",color="darkblue")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=df,x="Rating",bins=30,kde=True)
plt.title("Distribution of Rating",
          fontsize=16,
          fontweight="bold",
          color="darkblue",
          pad=20)
plt.xlabel("Rating",fontsize=12,fontweight="bold",color="darkblue")
plt.ylabel("Count",fontsize=12,fontweight="bold",color="darkblue")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=df,x="Number_of_Reviews",bins=20,kde=True)
plt.title("Distribution of Number of Reviews",
          fontsize=16,
          fontweight="bold",
          color="darkblue",
          pad=20)
plt.xlabel("Number of Reviews",fontsize=12,fontweight="bold",color="darkblue")
plt.ylabel("Count",fontsize=12,fontweight="bold",color="darkblue")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df,x="Price_USD",y="Rating",hue="Gender_Target")
plt.title("Price vs Rating by Gender",
          fontsize=16,
          fontweight="bold")
plt.xlabel("Price in USD",fontsize=12,fontweight="bold")
plt.ylabel("Rating",fontsize=12,fontweight="bold")
plt.legend(loc="upper right")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df,x="Number_of_Reviews",y="Rating",hue="Gender_Target")
plt.title("Number of Reviews vs Rating by Gender",
          fontsize=16,
          fontweight="bold")
plt.xlabel("Number of Reviews",fontsize=12,fontweight="bold")
plt.ylabel("Rating",fontsize=12,fontweight="bold")
plt.legend(loc="upper right")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x="Gender_Target",
    y="Rating"
)

plt.title("Rating Distribution by Gender")
plt.show()
```

```python
category_rating = (
    df.groupby("Category")["Rating"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12,6))

sns.barplot(
    x=category_rating.index,
    y=category_rating.values
)

plt.title("Average Rating by Product Category",
          fontsize=16,
          fontweight="bold")

plt.xlabel("Category")
plt.ylabel("Average Rating")

plt.xticks(rotation=45)
plt.show()
```

```python
skin_rating = (
    df.groupby("Skin_Type")["Rating"]
    .mean()
    .sort_values(ascending=False)
)
plt.figure(figsize=(10,6))
sns.barplot(
    x=skin_rating.index,
    y=skin_rating.values
)
plt.title("Average Rating by Skin Type",
          fontsize=16,
          fontweight="bold")
plt.xlabel("Skin Type")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.show()
```

```python
brand_rating=(
    df.groupby("Brand")["Rating"].mean().sort_values(ascending=False).head(10)
)
plt.figure(figsize=(12,6))
sns.barplot(x=brand_rating.values,y=brand_rating.index)
plt.title("Average Rating by Brand",
          fontsize=16,
          fontweight="bold")
plt.xlabel("Average Rating")
plt.ylabel("Brand")
plt.xticks(rotation=45)
plt.show()
```

```python
brand_category=df.pivot_table(
    values="Rating",
    index="Brand",
    columns="Category",
    aggfunc="mean"
)
plt.figure(figsize=(12,8))
sns.heatmap(brand_category, annot=True, cmap="YlGnBu",fmt=".2f")
plt.title("Average Rating by Brand and Category",
          fontsize=16,
          fontweight="bold")
plt.xlabel("Category")
plt.ylabel("Brand")
plt.show()
```

```python
skin_category=df.pivot_table(
    values="Rating",
    index="Skin_Type",
    columns="Category",
    aggfunc="mean"
)
plt.figure(figsize=(12,8))
sns.heatmap(skin_category, annot=True, cmap="YlGnBu",fmt=".2f")
plt.title("Average Rating by Skin Type and Category",
            fontsize=16,
            fontweight="bold")
plt.xlabel("Category")
plt.ylabel("Skin Type")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df,x="Price_USD",y="Rating",hue="Gender_Target")
plt.title("Price vs Rating by Gender", fontsize=16, fontweight="bold")
plt.xlabel("Price in USD", fontsize=12, fontweight="bold")
plt.ylabel("Rating", fontsize=12, fontweight="bold")
plt.legend(loc="upper right")
plt.show()
```

```python
brand_analysis = df.groupby("Brand").agg(
    Average_Price=("Price_USD", "mean"),
    Average_Rating=("Rating", "mean"),
    Total_Reviews=("Number_of_Reviews", "sum")
).reset_index()

print(brand_analysis.head(10))
```

```python
ingredient_category = df.pivot_table(
    values="Rating",
    index="Main_Ingredient",
    columns="Category",
    aggfunc="mean"
)

plt.figure(figsize=(15,8))

sns.heatmap(
    ingredient_category,
    annot=True,
    cmap="YlOrRd",
    fmt=".2f"
)

plt.title("Average Rating by Ingredient and Category",
          fontsize=16,
          fontweight="bold")

plt.xlabel("Category")
plt.ylabel("Main Ingredient")

plt.show()
```

```python
numerical_cols = ["Price_USD", "Rating", "Number_of_Reviews"]

correlation = df[numerical_cols].corr()

print(correlation)
plt.figure(figsize=(10,7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Features",
          fontsize=16,
          fontweight="bold")

plt.show()
```

```python
print("Price vs Rating:",
      df["Price_USD"].corr(df["Rating"]))

print("\nReviews vs Rating:",
      df["Number_of_Reviews"].corr(df["Rating"]))

print("\nPrice vs Reviews:",
      df["Price_USD"].corr(df["Number_of_Reviews"]))
```

```python
correlation_values = pd.Series({
    "Price vs Rating": df["Price_USD"].corr(df["Rating"]),
    "Reviews vs Rating": df["Number_of_Reviews"].corr(df["Rating"]),
    "Price vs Reviews": df["Price_USD"].corr(df["Number_of_Reviews"])
})

plt.figure(figsize=(10,6))

sns.barplot(
    x=correlation_values.index,
    y=correlation_values.values
)

plt.title("Correlation Between Numerical Features",
          fontsize=16,
          fontweight="bold")

plt.xlabel("Variable Pairs")
plt.ylabel("Correlation Coefficient")

plt.axhline(0, color="black", linewidth=1)
plt.xticks(rotation=15)

plt.show()
```
