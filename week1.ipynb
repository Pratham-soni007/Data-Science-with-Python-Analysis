```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv("most_used_beauty_cosmetics_products_extended.csv")
data.head(5)
```

```python
plt.figure(figsize=(26,8))
sns.barplot(x="Category",y="Rating",
            data=data,
            hue="Gender_Target",
            palette="cubehelix",
            legend=True)
plt.title("Category v/s Rating",
          fontsize=16,
          fontweight="bold",
          fontfamily="Arial",
          fontstyle="italic",)
plt.show()
```

```python
plt.figure(figsize=(53,20))
sns.barplot(x="Brand",y="Rating",
            data=data,
            palette="crest",
            hue="Gender_Target",
            legend=True)
plt.title("Brand v/s Rating",
          fontsize=20,
          fontweight="bold",
          fontfamily="Arial")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x="Price_USD",
                y="Rating",
                data=data,
                color="royalblue",
                alpha=0.7,
                s=15
                )
plt.title("Product Price v/s Rating",
          fontsize=16,
          fontweight="bold")
plt.xlabel("Price (USD)",
           fontsize=12,
           fontweight="bold")
plt.ylabel("Rating",
           fontsize=16,
           fontweight="bold")
plt.tight_layout()
plt.show()
correlation=data["Price_USD"].corr(data["Rating"])
print("Price-Rating Correlation Value: ",round(correlation,3))
if correlation >=0.7:
    print("Strong Positive Correlation")
elif correlation >=0.3:
    print("Moderate Positive Correlation")
elif correlation >=0:
    print("Weak Positive Correlation")
elif correlation <=-0.7:
    print("Strong Negative Correlation")
elif correlation <=-0.3:
    print("Moderate Negative Correlation")
elif correlation <0:
    print("Weak Negative Correlation")
else:
    print("No linear Relationship")
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x="Number_of_Reviews",
            y="Rating",
            data=data,
            color="red",
            alpha=0.7,
            )
plt.title("Number of Reviews v/s Rating",
          fontsize=16,
          fontweight="bold",
          fontfamily="Arial")
plt.tight_layout()
plt.show()
correlation=data["Number_of_Reviews"].corr(data["Rating"])
print("Correlation Value: ",round(correlation,3))
if correlation >=0.7:
    print("Strong Positive Correlation")
elif correlation >=0.3:
    print("Moderate Positive Correlation")
elif correlation >=0:
    print("Weak Positive Correlation")
elif correlation <=-0.7:
    print("Strong Negative Correlation")
elif correlation <=-0.3:
    print("Moderate Negative Correlation")
elif correlation <0:
    print("Weak Negative Correlation")
else:
    print("No linear Relationship")
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(
    x="Skin_Type",
    y="Rating",
    data=data,
    hue="Gender_Target",
    palette="flare",
    legend=True
)
plt.title("Skin_Type v/s Rating",
          fontsize=16,
          fontweight="bold",
          fontfamily="Arial"
          )
plt.tight_layout()
plt.legend(loc="upper right")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(y="Main_Ingredient",
            x="Rating",
            data=data)
plt.title("Main Ingredient v/s Rating",
          fontsize=16,
          fontweight="bold",
          fontfamily="Arial")
plt.show()
```

```python
data["Cruelty_Status"]=data["Cruelty_Free"].map({True:"Cruelty-Free",False:"Non-Cruelty-Free"})
fig,axes=plt.subplots(1,2,figsize=(12,5))
sns.boxplot(data=data,
            x="Cruelty_Status",
            y="Price_USD",
            ax=axes[0]
            )
axes[0].set_title("Price by Cruelty Status")
sns.boxplot(data=data,
            x="Cruelty_Status",
            y="Rating",
            ax=axes[1]
            )
axes[1].set_title("Ratings by Cruelty Status")
plt.tight_layout()
plt.show()
```
